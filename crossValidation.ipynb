### Cross-validation motivation


>Model performance is dependent on the way we split up the data
>Not representative of the model's ability to generalize to unseen data
>Solution: Cross-validation!

Cross-validation and model performance
5 folds = 5-fold CV
10 folds = 10-fold CV
k folds = k-fold CV
More folds = More computationally expensiv

```python
import pandas as pd
import numpy as np
diabetes_df=pd.read_csv(r"C:\Users\user\Downloads\diabetes_prediction_dataset.csv\diabetes_prediction_dataset.csv")
import pandas as pd
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

diabetes_df["gender"] = le.fit_transform(diabetes_df["gender"])#label encoding

diabetes_df = pd.get_dummies(
    diabetes_df,
    columns=["smoking_history"],
    dtype=int
)#one_hot encoding
```

```python
X=diabetes_df.drop(["blood_glucose_level"],axis=1).values
y=diabetes_df["blood_glucose_level"].values
```

# Cross-validation in scikit-learn

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score, KFold 
kf = KFold(n_splits=6, shuffle=True, random_state=42) 
reg = LinearRegression() 
cv_results = cross_val_score(reg, X, y, cv=kf)
```

```python
#Evaluating cross-validation peformance
print(cv_results)
```

```python
print(np.mean(cv_results), np.std(cv_results))
```

```python
print(np.quantile(cv_results, [0.025, 0.975]))
```
